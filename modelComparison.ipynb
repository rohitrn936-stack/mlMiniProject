```python
import pandas as pd

comparisonData = pd.DataFrame({
    "model": [
        "Naive Bayes",
        "Logistic Regression"
    ],
    "validationAccuracy": [
        0.50,
        0.5833333333333334
    ],
    "predictedMVP": [
        "Nikola Jokić",
        "Nikola Jokić"
    ]
})

print(comparisonData)
```

```python
bestModel = comparisonData.loc[
    comparisonData["validationAccuracy"].idxmax(),
    "model"
]

bestAccuracy = comparisonData["validationAccuracy"].max()

print("Best Model:", bestModel)
print("Best Validation Accuracy:", bestAccuracy)
```
