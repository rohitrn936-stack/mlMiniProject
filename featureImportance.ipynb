```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

trainingData = pd.read_csv("NBA_MVP_Training.csv")

features = [
    "games",
    "minutesPerGame",
    "fieldGoalPercentage",
    "threePointPercentage",
    "freeThrowPercentage",
    "reboundsPerGame",
    "assistsPerGame",
    "stealsPerGame",
    "blocksPerGame",
    "pointsPerGame"
]

xTrain = trainingData[features]
yTrain = trainingData["mvp"]

model = LogisticRegression(max_iter=1000)

model.fit(xTrain, yTrain)

featureData = pd.DataFrame({
    "feature": features,
    "coefficient": model.coef_[0]
})

featureData["importance"] = featureData["coefficient"].abs()

featureData = featureData.sort_values(
    "importance",
    ascending=False
)

print(featureData)
```
