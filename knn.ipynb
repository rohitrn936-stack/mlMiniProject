```python
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
trainingData = pd.read_csv("NBA_MVP_Training.csv")
validationData = pd.read_csv("NBA_MVP_Validation.csv")
predictionData = pd.read_csv("NBA_MVP_2021.csv")

features = ['games', 
            'minutesPerGame', 
            'fieldGoalPercentage', 
            'threePointPercentage', 
            'freeThrowPercentage', 
            'reboundsPerGame', 
            'assistsPerGame', 
            'stealsPerGame', 
            'blocksPerGame', 
            'pointsPerGame']

xTrain = trainingData[features]
yTrain = trainingData["mvp"]
xValidation = validationData[features]
yValidation = validationData["mvp"]

model = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5)
)
model.fit(xTrain, yTrain)

validationPrediction = model.predict(xValidation)
accuracy = accuracy_score(yValidation, validationPrediction)
precision = precision_score(yValidation, validationPrediction, zero_division=0)
recall = recall_score(yValidation, validationPrediction, zero_division=0)

print("KNN Validation Accuracy:", accuracy)
print("KNN Validation Precision:", precision)
print("KNN Validation Recall:", recall)
```

```python
xPrediction = predictionData[features]
predictionData["prediction"] = model.predict(xPrediction)
predictionData["probability"] = model.predict_proba(xPrediction)[:, 1]
print(predictionData[["player", "prediction", "probability"]])

bestPlayer = predictionData.loc[
    predictionData["probability"].idxmax(),
    "player"
]
print("\\nPredicted 2021 MVP:", bestPlayer)
```
